### promts templates - dynamic prompts


```python
import os
from dotenv import load_dotenv
from langchain_nvidia_ai_endpoints import ChatNVIDIA
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


load_dotenv(override=True)
out = StrOutputParser()

api_key = os.getenv("NEMOTRON_NVIDIA_API_KEY")
llm = ChatNVIDIA(model = "nvidia/nemotron-3-ultra-550b-a55b",
nvidia_api_key=api_key,)
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role":"system" , "content":"translate the input in {language}"},
   {"role":"user" , "content":"{query}"}
])
```

```python
def transformStr(result:str):
    return result.upper()
```

```python
final_prompt = prompts.format_messages(language = 'hinglish',
                                        query = 'i dont want to do work')

res = llm.invoke(final_prompt)
print(res.content)                                      
```
