#### connect with google geni mode


```python
import os
import time
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv(override=True)

api_key = os.getenv("GOOGLE_API_KEY")
assert api_key, "GOOGLE_API_KEY was not loaded from .env"

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=api_key,
    timeout=60,
    max_retries=1,
)

response = llm.invoke("who is virat kohli")
print(response.content)
```

### Static Prompt
#### here system prompt is fixed and user promt also fixed

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=api_key
)
```

```python
prompts = [
    ("system", "you are javascript developer"),
    ("user", "how to sort an array")
]
res = llm.invoke(prompts)
print(res.content[0]["text"])
```
